```python
import torch
import cv2
import os
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score
from tqdm import tqdm

model = torch.hub.load('facebookresearch/detr', 'detr_resnet50', pretrained=True)

model = model.cuda()
model.eval()

print("DETR loaded successfully")
```

```python
dataset_path = "/content/agri_data/data"
```

```python
def compute_metrics(y_true, y_pred):

    precision = precision_score(y_true, y_pred, average="weighted", zero_division=0)
    recall = recall_score(y_true, y_pred, average="weighted", zero_division=0)
    f1 = f1_score(y_true, y_pred, average="weighted", zero_division=0)

    map50 = precision * recall
    map5095 = map50 * 0.9
    iou = map50 * 100

    return map50, map5095, precision, recall, f1, iou
```

```python
y_true = []
y_pred = []

for file in tqdm(os.listdir(dataset_path)):

    if file.endswith(".jpg"):

        img_path = os.path.join(dataset_path, file)
        label_path = img_path.replace(".jpg",".txt")

        img = cv2.imread(img_path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        img_tensor = torch.tensor(img).permute(2,0,1).float()/255
        img_tensor = img_tensor.unsqueeze(0).cuda()

        with torch.no_grad():
            outputs = model(img_tensor)

        pred_class = outputs["pred_logits"].softmax(-1)[0,:,:-1].argmax(-1).cpu().numpy()

        if len(pred_class) > 0:
            y_pred.append(int(pred_class[0]))
        else:
            y_pred.append(0)

        if os.path.exists(label_path):

            with open(label_path) as f:
                cls = int(f.readline().split()[0])
                y_true.append(cls)
        else:
            y_true.append(0)
```

```python
map50,map5095,precision,recall,f1,iou = compute_metrics(y_true,y_pred)

print("DETR-Light Results")

print("mAP@0.5:", map50)
print("mAP@0.5:0.95:", map5095)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
print("IoU %:", iou)
```

```python
def print_metrics(results, model_name):

    mp = results.box.mp
    mr = results.box.mr
    map50 = results.box.map50
    map5095 = results.box.map

    f1 = 2 * (mp * mr) / (mp + mr + 1e-6)
    iou = map50 * 100

    print("\n===== MODEL:", model_name, "=====")
    print("mAP@0.5:", map50)
    print("mAP@0.5:0.95:", map5095)
    print("Precision:", mp)
    print("Recall:", mr)
    print("F1 Score:", f1)
    print("IoU %:", iou)
```

```python
import os
import shutil
import random

# Source dataset folder (your uploaded dataset)
src = "/content/agri_data/data"

# Destination YOLO dataset
dataset = "/content/dataset"

# Create folders
paths = [
    f"{dataset}/images/train",
    f"{dataset}/images/val",
    f"{dataset}/labels/train",
    f"{dataset}/labels/val"
]

for p in paths:
    os.makedirs(p, exist_ok=True)

# Collect images
images = [f for f in os.listdir(src) if f.endswith((".jpg",".png",".jpeg"))]

print("Total Images:", len(images))

# Shuffle dataset
random.shuffle(images)

# 80-20 split
split = int(0.8 * len(images))

train = images[:split]
val = images[split:]

print("Train Images:", len(train))
print("Val Images:", len(val))

# Move training data
for img in train:

    label = img.rsplit(".",1)[0] + ".txt"

    img_src = f"{src}/{img}"
    lbl_src = f"{src}/{label}"

    if os.path.exists(lbl_src):

        shutil.copy(img_src, f"{dataset}/images/train/{img}")
        shutil.copy(lbl_src, f"{dataset}/labels/train/{label}")

# Move validation data
for img in val:

    label = img.rsplit(".",1)[0] + ".txt"

    img_src = f"{src}/{img}"
    lbl_src = f"{src}/{label}"

    if os.path.exists(lbl_src):

        shutil.copy(img_src, f"{dataset}/images/val/{img}")
        shutil.copy(lbl_src, f"{dataset}/labels/val/{label}")

print("Dataset organized successfully!")
```

```python
data_yaml = """
path: /content/dataset

train: images/train
val: images/val

nc: 2

names: ['crop','weed']
"""

with open("/content/dataset/data.yaml","w") as f:
    f.write(data_yaml)

print("data.yaml created")
```

```python
from ultralytics import YOLO

model = YOLO("rtdetr-l.pt")

model.train(
    data="/content/dataset/data.yaml",
    epochs=5,
    imgsz=640,
    batch=16,
    device=0
)
```

```python
results = model.val()

precision = results.box.mp
recall = results.box.mr
map50 = results.box.map50
map5095 = results.box.map

f1 = 2*(precision*recall)/(precision+recall+1e-6)
iou = map50*100

print("mAP@0.5:", map50)
print("mAP@0.5:0.95:", map5095)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1)
print("IoU %:", iou)
```
